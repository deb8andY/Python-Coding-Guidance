### **`Python Local & Global Variables Declaration Techniques - `**
*   Local variable names start with 'v_'
*   Global variable names start with 'gv_'

```python
v_varName = 'local value'
gv_varName = 'global value'
print(f"v_varName: {v_varName}, gv_varName: {gv_varName}")
```

### **Python string variable concat options in print statements** -
*   Using ','
*   Using F function
*   Using .format() function

```python
name="abigail"
print("name is: ",name)
```

```python
name="john"
age=35
print(f"name: {name}, age: {str(age)}")
```

```python
print("name: {}, age: {}".format(name, str(age)))
```

```python
print("name: {1}, age: {0}".format(name, str(age)))
```

```python
print("name: {b}, age: {a}".format(b=name, a=str(age)))
```
